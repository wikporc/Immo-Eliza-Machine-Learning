```python
import json
import joblib
from pathlib import Path

# -----------------------------------
# CONFIG
# -----------------------------------
MODEL_PATH = Path(r"models_optimisation\cat_res_apt_BEST.pkl")
OUTPUT_JSON = Path(r"models_optimisation\best_parameters\cat_res_apt_best_params.json")

# -----------------------------------
# LOAD
# -----------------------------------
model_loaded = joblib.load(MODEL_PATH)

if hasattr(model_loaded, "named_steps"):
    model = model_loaded.named_steps["model"]
else:
    model = model_loaded

# -----------------------------------
# EXTRACT
# -----------------------------------
PARAM_KEYS = [ "depth", "learning_rate", "l2_leaf_reg", "bagging_temperature", "random_strength", "iterations" ]

params = {}
for key in PARAM_KEYS:
    try:
        params[key] = model.get_params()[key]
    except:
        params[key] = None

params["objective"] = "regression"
params["metric"] = "rmse"
params["verbosity"] = -1
params["force_col_wise"] = True

with open(OUTPUT_JSON, "w") as f:
    json.dump(params, f, indent=4)

print(f"Saved params to {OUTPUT_JSON.resolve()}")
print("Parameters:", params)
```

```python
import json
import joblib
from pathlib import Path

# -----------------------------------
# CONFIG
# -----------------------------------
MODEL_PATH = Path(r"models_optimisation\xgb_res_apt_BEST.pkl")
OUTPUT_JSON = Path(r"models_optimisation\best_parameters\xgb_res_apt_best_params.json")

# -----------------------------------
# LOAD MODEL
# -----------------------------------
model_loaded = joblib.load(MODEL_PATH)

if hasattr(model_loaded, "named_steps"):
    model = model_loaded.named_steps["model"]
else:
    model = model_loaded

# -----------------------------------
# EXTRACT PARAMS
# -----------------------------------
PARAM_KEYS = [
    "max_depth",
    "learning_rate",
    "subsample",
    "colsample_bytree",
    "min_child_weight",
    "n_estimators"
]

params = {}
for key in PARAM_KEYS:
    try:
        params[key] = model.get_params()[key]
    except:
        params[key] = None

params["objective"] = "reg:squarederror"
params["tree_method"] = "hist"
params["verbosity"] = 0

with open(OUTPUT_JSON, "w") as f:
    json.dump(params, f, indent=4)

print(f"Saved params to {OUTPUT_JSON.resolve()}")
print("Parameters:", params)
```
