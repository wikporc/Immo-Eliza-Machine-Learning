This notebook extracts raw features used by the models for Apartments and houses and saves them into .json files.

```python
import joblib
import json

pipeline = joblib.load(r"models_optimisation\optimum_models\model_cat_res_apt.pkl")

preproc = pipeline.named_steps["preprocess"]

raw_features = []

for name, transformer, cols in preproc.transformers_:
    if name == "remainder" and transformer == "drop":
        continue
    
    raw_features.extend(list(cols))


with open(r"models_optimisation\raw_features\raw_features_apt.json", "w") as f:
    json.dump(raw_features, f, indent=2)

print("RAW FEATURES:", raw_features)
print("Saved as raw_features_apt.json")
```
